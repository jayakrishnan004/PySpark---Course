PySpark SQL is a module in Apache Spark that allows us to process and analyze structured data using SQL queries and DataFrame operations.

```python
import pyspark
from pyspark.sql import SparkSession
spark = SparkSession.builder.getOrCreate()
spark
```

```python
taxi = spark.read.parquet("/content/drive/MyDrive/DSA - ICTAK/PySpark/Dataset/yellow_tripdata_2025-01.parquet")
```

```python
taxi.createOrReplaceTempView("taxi")
# registers the dataframe with the spark SQL catalog as a temporary view.
```

```python
spark.sql("SELECT * FROM TAXI WHERE TOTAL_AMOUNT > 0").show()
```

```python
# Display payment_type, passenger_count, and total_amount for taxi trips where the total amount exceeds 50 and the passenger count is greater than 2.
spark.sql("SELECT * FROM TAXI WHERE TOTAL_AMOUNT > 50").filter("passenger_count > 2").select("payment_type", "passenger_count", "total_amount").show()
```

```python
# Fully using SQL query.
q = """select payment_type, passenger_count, total_amount from taxi where total_amount > 50 and passenger_count > 2"""
```

```python
spark.sql(q).show()
```

# **CHALLENGE**
* Load the January taxi ride data into a dataframe called taxi_jan2025 and register a temporary view with the same name.
* Load the taxi zone lookup data into a dataframe called taxi_lookup and register a temporary view with the same name.
* Use PySpark SQL to left join the two tables on the DOLocationID and the LocationID columns.
 - * Only select the DOLocationID, Borough, and total_amount columns.
 - * Assign the result to a dataframe named joined_df.
 * Using the PySpark dataframe syntax, group the result by the Borough column and calculate the average total_amount using the avg method.
- * Alias the average column as avg_amount
- * Display the resulting dataframe in the notebook

```python
# Load the January taxi ride data into a dataframe called taxi_jan2025 and register a temporary view with the same name.
taxi_jan_2025 = spark.read.parquet("/content/drive/MyDrive/DSA - ICTAK/PySpark/Dataset/yellow_tripdata_2025-01.parquet")
taxi_jan_2025.createOrReplaceTempView("taxi_jan_2025")
```

```python
# Load the taxi zone lookup data into a dataframe called taxi_lookup and register a temporary view with the same name.
taxi_lookup = spark.read.option("header", "true").csv("/content/drive/MyDrive/DSA - ICTAK/PySpark/Dataset/taxi_zone_lookup.csv")
taxi_lookup.createOrReplaceTempView("taxi_lookup")
```

```python
# Use PySpark SQL to left join the two tables on the DOLocationID and the LocationID columns.
# Only select the DOLocationID, Borough, and total_amount columns.
# Assign the result to a dataframe named joined_df.
q = """select t1.DOLocationID, t2.Borough, t1.total_amount from taxi_jan_2025 t1 left join taxi_lookup t2 on t1.DOLocationID = t2.LocationID"""
joined_df = spark.sql(q)
joined_df.show()
```

```python
# Using the PySpark dataframe syntax, group the result by the Borough column and calculate the average total_amount using the avg method.
# Alias the average column as avg_amount.
# Display the resulting dataframe in the notebook.

from pyspark.sql.functions import avg
joined_df.groupBy("Borough").agg(avg("total_amount").alias("avg_amount")).show()
```
