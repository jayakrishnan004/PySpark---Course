# **HANDLING MISSING DATA**

```python
from pyspark.sql.functions import col, isnull
```

```python
df.filter(isnull(col("fare_amount"))).count()
```

```python
# Find the null values.
df.filter(isnull(col("passenger_count"))).count()
```

```python
# Filling the missing values.
df1 = df.fillna({"passenger_count":1})
df.show()
```

```python
df.filter(isnull(col("fare_amount"))).count()
```

```python
# Create a new column showing trip duration in minutes.
# withcolumn()

from pyspark.sql.functions import unix_timestamp, round

df1 = df.withColumn("trip_duration_minutes", round((unix_timestamp("tpep_dropoff_datetime") - unix_timestamp("tpep_pickup_datetime"))/60,1))
df1.show()
```

```python
# Rename column.
df2 = df1.select("tpep_pickup_datetime","tpep_dropoff_datetime",\
                 "PULocationID","DOLocationID","trip_duration_minutes",\
                 "passenger_count","fare_amount","Airport_fee","total_amount")\
.withColumnsRenamed({"tpep_pickup_datetime" : "pu_datetime",\
                    "tpep_dropoff_datetime" : "do_datetime",\
                     "PULocationID" : "pu_loc_id",\
                     "DOLocationID" : "do_loc_id",\
                     "Airport_fee" : "airport_fee"
                     })
```

```python
df2.show()
```

```python
# Removes columns from a dataframe.
df1.drop("VendorID", "RatecodeID").show()
```

# **AGGREGATE FUNCTIONS**

Data Dictionary for payment lookup - payment_type.
* 0 = Flex fare trip
* 1 = credit card payment
* 2 = cash payment
* 3 = no charge
* 4 = dispute
* 5 = unknown
* 6 = voided trip

```python
df.groupby("payment_type").count().sort("payment_type").show()
```

```python
df.groupby("payment_type").avg("total_amount").show()
```

```python
from pyspark.sql.functions import avg, sum
df.groupBy("payment_type").agg(avg("total_amount")).show()
```

```python
df.groupBy("payment_type").agg(avg("total_amount").alias("avg_amount")).show()
```

```python
df.groupBy("passenger_count").agg(\
    sum("total_amount").alias("total_revenue"),\
    avg("total_amount").alias("average_fare")\
).show()
```

## **CHALLENGE**
create two new dataframes called df_jan_2025 and df_feb_2025 from the corresponding data files.
* create a new dataframe called df_2025_combined as union of those two files.
*  Only select the following columns from this combined dataframe and rename columns as indicated in parentheses. Reassign the result to df_2025_combined.

tpep_pickup_datetime (pu_datetime), tpep_dropoff_datetime (do_datetime), PULocationID (pu_location_id), DOLocationID (do_location_id), passenger_count, fare_amount, Airport_fee (airport_fee), total_amount, payment_type
* create a new dataframe called taxi_zones from the taxi_zone_lookup.csv file.
* Join the taxi_zones dataframe onto the df_2025_combined dataframe using the do_location_id and LocationID columns. Reassign the result of this join to df_2025_combined.
* Final cleanup: Drop the superfluous LocationID, Zone, and service_zone columns, and rename the Borough column to pu_boro. Reassign the result to df_2025_combined.
* display the df_2025_combined dataframe.

```python
# create two new dataframes called df_jan_2025 and df_feb_2025 from the corresponding data files.
df_jan_2025 = spark.read.parquet("/content/drive/MyDrive/DSA - ICTAK/PySpark/Dataset/yellow_tripdata_2025-01.parquet")
df_feb_2025 = spark.read.parquet("/content/drive/MyDrive/DSA - ICTAK/PySpark/Dataset/yellow_tripdata_2025-02.parquet")
```

```python
# create a new dataframe called df_2025_combined as union of those two files.
df_2025_combined = df_jan_2025.union(df_feb_2025)
```

```python
# Only select the following columns from this combined dataframe and rename columns as indicated in parentheses. Reassign the result to df_2025_combined.
# if there occurs an error, re run the above two cells and run this cell again.

df_2025_combined = df_2025_combined.select("tpep_pickup_datetime", "tpep_dropoff_datetime", "PULocationID", "DOLocationID", "passenger_count", "fare_amount", "Airport_fee", "total_amount", "payment_type").withColumnsRenamed({"tpep_pickup_datetime" : "pu_datetime","tpep_dropoff_datetime" : "do_datetime", "PULocationID" : "pu_location_id", "DOLocationID": "do_location_id"})
```

```python
df_2025_combined.show()
```

```python
# create a new dataframe called taxi_zones from the taxi_zone_lookup.csv file.
taxi_zones = spark.read.option("header", "True").csv("/content/drive/MyDrive/DSA - ICTAK/PySpark/Dataset/taxi_zone_lookup.csv")
```

```python
# Join the taxi_zones dataframe onto the df_2025_combined dataframe using the do_location_id and LocationID columns. Reassign the result of this join to df_2025_combined.
df_2025_combined = df_2025_combined.join(taxi_zones, df_2025_combined.do_location_id == taxi_zones.LocationID, "left")
df_2025_combined.show()
```

```python
# Drop the superfluous LocationID, Zone, and service_zone columns, and rename the Borough column to pu_boro. Reassign the result to df_2025_combined.
df_2025_combined = df_2025_combined.drop("LocationID", "zone", "service_zone").withColumnsRenamed({"Borough" : "pu_boro"})
df_2025_combined.show()
```
